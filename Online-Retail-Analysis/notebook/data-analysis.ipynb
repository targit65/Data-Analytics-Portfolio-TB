```python
# Category-level Sales By Revenue
category_price = (df.groupby(['category'], as_index=False
                          ).agg(avg_price = ('price', 'mean'),
                                units_sold = ('sold_count', 'sum'),
                                revenue = ('revenue', 'sum')
                               ).sort_values('revenue', ascending=False)
                )
category_price.head(10)
```

```python
# Category-level Sales by Average price(taking all prices including extreme high prices)
category_price = (df.groupby(['category'], as_index=False
                          ).agg(avg_price = ('price', 'mean'),
                                units_sold = ('sold_count', 'sum'),
                                revenue = ('revenue', 'sum')
                               ).sort_values('avg_price', ascending=False)
                )
category_price.head(10)
```

```python
# Average price of category (high prices excluded)
df_price = df[df['price_placeholder'] == 0].copy()
category_price = (df_price.groupby(['category'], as_index=False
                          ).agg(avg_price = ('price', 'mean'),
                                units_sold = ('sold_count', 'sum'),
                                revenue = ('revenue', 'sum')
                               ).sort_values('avg_price', ascending=False)
                )
category_price.head(10)
```

```python
analysis_date = df["item_creation_date"].max()

df["listing_age_days"] = (
    analysis_date - df["item_creation_date"]
).dt.days
```

```python
df["listing_age_group"] = pd.cut(
    df["listing_age_days"],
    bins=[-1, 90, 180, 365, 730, float("inf")],
    labels=[
        "0-3 months",
        "3-6 months",
        "6-12 months",
        "1-2 years",
        "2+ years"
    ]
)
```

```python
age_sales = df.groupby(
    "listing_age_group",
    observed=True,
    as_index=False
)
age_sales = age_sales.agg(
    listings=("itemid", "nunique"),
    units_sold=("sold_count", "sum"),
    revenue=("revenue", "sum")
)
```
